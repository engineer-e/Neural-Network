### Weight Matrix

1. Typically, a neuron has more than one input. 
2. A neuron with *R* inputs is shown in Figure 2.5.
3. The individual inputs $p_1$,$p_2$,$\cdots$,$p_R$ are each weighted by corresponding elements $w_{1,1}$, $w_{1,2}$,$\cdots$ ,$w_{1,R}$ of the *weight matrix* **W**.
4. The neuron has a bias *b*, which is summed with the weighted inputs to form the net input *n*: 
$$\displaystyle n = w_{1,1}p_1 + w_{1,2}p_2 + \cdots + w_{1,R}p_R + b.$$
5. This expression can be written in matrix form:
$$\displaystyle n = Wp + b$$
6. where the matrix **W** for the single neuron case has only one row.
7. Now the neuron output can be written as 
$$\displaystyle a = f(Wp+b) $$
8. Fortunately, neural networks can often be described with matrices.
9. This kind of matrix expression will be used throughout the book. 
10. Don't be concerned if you are rusty with matrix and vector operations.
11. We will review these topics in Chapters 5 and 6, we will provide many examples and solved problems that will spell out the procedures.

```python
import numpy as np
weights = np.array([3])
inputs = np.array([2])
bias = -1.5
net_output = weights.dot(inputs) + bias
net_output
```

### Weight Indices

1. We have adopted a particular convention in assigning the indices of the elements of the weight matrix.
2. The first index indicates the particular neuron destination for that weight.
3. The second index indicates the source of the signal fed to the neuron.
4. Thus, the indices in $w_{1,2}$ say that this weight represents the connection to the first (and only) neuron  *from* the second source.
5. Of course, this convention is more useful if there is more than one neuron, as will be the case later in this chapter.

### Abbreviated Notation

1. We would like to draw networks with several neurons, each having several inputs.
2. Further, we would like to have more than one layer of neurons.
3. You can imagine how complex such a network might appear if all the lines were drawn.
4. It would take a lot of ink, could hardly to read, and the mass of detail might obscure the main features.
5. Thus, we will use an *abbreviated notation.* 
6. A multiple-input neuron using this notation is shown in Figure 2.6.
7. As shown in Figure 2.6, the input vector **p** is represented by the solid vertical bar at the left. 
8. The dimensions of **p** are displayed below the variable as $R × 1$, indicating that th input is a single vector of *R* elements.
9. These inputs go to the weight matrix **W**, which has *R* columns but only one row in this single neuron case.
10. A constant 1 enters the neuron as an input and is multiplied by a scalar bias *b*.
11. The net input to the transfer function *f* is *n*, which is the sum of the bias *b* and the product *Wp*. 
12. The neuron's output *a* is a scalar inth is case. 
13. If we had more than one neuron, the network output would be a Vector 
14. The dimensions of the variables is these abbreviated notation figures will always be included, so that you can tell immediately if we are talking about a scalar, a vector or a matrix. 
15. You will not have to guess the kind of variable or its dimensions.
16. Note that the number of inputs to a network is set by the external specifications of the problem. 
17. If, for instance, you want to design a neural network that is to predict kite-flying conditions and the inputs are air temperature, wind velocity, and there would be three inputs to the network.
