# Neuron Model

## Single-Input Neuron


> Weight, Bias, Net Input, Transfer Function

1. The scalar input _p_ is multiplied by the scalar weight _w_ to form _wp_, one of terms that is sent to the summar.
2. The other input, 1, is multiplied by a _bias b_ and then passed to the summer.
3. The Summer output _n_, often referred to as the _net input_, goes into a _transfer function f_, which produces the scalar neuron output _a_.
4. Some authors use the term
   - "activation function" rather than _transfer function_
   - "offset" rather than _bias_
5. This simple model back to the biological neuron that we discussed in _Chatper 1_
6. The weight _w_ corresponds to the strength of a synapse.
7. The Cell body is represented by the summation and the transfer function, and the neuron output _a_ represents the signal on the axon


$$\displaystyle a = f(wp+b)$$

```python
import ipywidgets as widgets
from IPython.display import display, clear_output
import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle, Circle, FancyArrowPatch


# ============================================================
# SINGLE-INPUT NEURON
#
# n = wp + b
# a = f(n)
#
# Transfer function:
# f(n) = n
# ============================================================

def f(n):
    return n


# ============================================================
# SLIDERS
# ============================================================

p_slider = widgets.FloatSlider(
    value=1.0,
    min=-5,
    max=5,
    step=0.1,
    description='Input p:',
    continuous_update=True,
    style={'description_width': '80px'},
    layout=widgets.Layout(width='400px')
)

w_slider = widgets.FloatSlider(
    value=1.0,
    min=-5,
    max=5,
    step=0.1,
    description='Weight w:',
    continuous_update=True,
    style={'description_width': '80px'},
    layout=widgets.Layout(width='400px')
)

b_slider = widgets.FloatSlider(
    value=0.0,
    min=-5,
    max=5,
    step=0.1,
    description='Bias b:',
    continuous_update=True,
    style={'description_width': '80px'},
    layout=widgets.Layout(width='400px')
)


# ============================================================
# DRAW NEURON
# ============================================================

def draw_neuron(p, w, b):

    # -----------------------------
    # Neuron calculation
    # -----------------------------

    n = w * p + b
    a = f(n)


    # -----------------------------
    # Figure
    # -----------------------------

    fig = plt.figure(
        figsize=(13, 7),
        facecolor="#F5F7FA"
    )


    # ========================================================
    # NEURON DIAGRAM
    # ========================================================

    ax = fig.add_axes([0.03, 0.15, 0.65, 0.72])

    ax.set_xlim(0, 12)
    ax.set_ylim(0, 8)

    ax.axis("off")


    # ========================================================
    # TITLE
    # ========================================================

    ax.text(
        6,
        7.5,
        "Single-Input Neuron",
        ha="center",
        fontsize=25,
        fontweight="bold",
        color="#17202A"
    )


    # ========================================================
    # INPUT p
    # ========================================================

    ax.text(
        0.8,
        5.2,
        "Input",
        fontsize=17,
        fontweight="bold",
        color="#2874A6"
    )

    ax.add_patch(
        Circle(
            (1.5, 4.2),
            0.13,
            color="#2874A6"
        )
    )

    ax.text(
        1.15,
        3.65,
        f"p = {p:.1f}",
        fontsize=20,
        color="#2874A6"
    )


    # ========================================================
    # INPUT ARROW
    # ========================================================

    ax.add_patch(
        FancyArrowPatch(
            (1.65, 4.2),
            (3.2, 4.2),
            arrowstyle="-|>",
            mutation_scale=22,
            linewidth=2.5,
            color="#2874A6"
        )
    )


    # ========================================================
    # WEIGHT
    # ========================================================

    ax.text(
        2.25,
        4.65,
        f"w = {w:.1f}",
        fontsize=18,
        color="#8E44AD"
    )


    # ========================================================
    # SUMMATION BLOCK
    # ========================================================

    ax.add_patch(
        Rectangle(
            (3.2, 3.55),
            1.2,
            1.3,
            linewidth=3,
            edgecolor="#2C3E50",
            facecolor="white"
        )
    )

    ax.text(
        3.8,
        4.2,
        r"$\Sigma$",
        ha="center",
        va="center",
        fontsize=36,
        color="#2C3E50"
    )


    # ========================================================
    # BIAS
    # ========================================================

    ax.add_patch(
        Circle(
            (3.8, 2.0),
            0.13,
            color="#E67E22"
        )
    )

    ax.add_patch(
        FancyArrowPatch(
            (3.8, 2.15),
            (3.8, 3.55),
            arrowstyle="-|>",
            mutation_scale=20,
            linewidth=2.5,
            color="#E67E22"
        )
    )

    ax.text(
        4.15,
        2.45,
        f"b = {b:.1f}",
        fontsize=18,
        color="#E67E22"
    )


    # ========================================================
    # n
    # ========================================================

    ax.add_patch(
        FancyArrowPatch(
            (4.4, 4.2),
            (5.7, 4.2),
            arrowstyle="-|>",
            mutation_scale=22,
            linewidth=2.5,
            color="#34495E"
        )
    )

    ax.text(
        5.0,
        4.65,
        f"n = {n:.2f}",
        fontsize=18,
        color="#34495E"
    )


    # ========================================================
    # TRANSFER FUNCTION
    # ========================================================

    ax.add_patch(
        Rectangle(
            (5.7, 3.55),
            1.3,
            1.3,
            linewidth=3,
            edgecolor="#27AE60",
            facecolor="#EAF8F0"
        )
    )

    ax.text(
        6.35,
        4.2,
        r"$f$",
        ha="center",
        va="center",
        fontsize=34,
        color="#27AE60"
    )


    # ========================================================
    # OUTPUT
    # ========================================================

    ax.add_patch(
        FancyArrowPatch(
            (7.0, 4.2),
            (8.6, 4.2),
            arrowstyle="-|>",
            mutation_scale=22,
            linewidth=2.5,
            color="#C0392B"
        )
    )

    ax.text(
        8.9,
        4.2,
        f"a = {a:.2f}",
        fontsize=22,
        fontweight="bold",
        color="#C0392B"
    )


    # ========================================================
    # EQUATION
    # ========================================================

    ax.text(
        6.0,
        1.2,
        r"$a=f(wp+b)=f(n)$",
        ha="center",
        fontsize=23,
        color="#17202A"
    )


    # ========================================================
    # CALCULATION PANEL
    # ========================================================

    calc = fig.add_axes(
        [0.71, 0.35, 0.26, 0.45]
    )

    calc.axis("off")


    calc.text(
        0.5,
        0.95,
        "Live Calculation",
        ha="center",
        fontsize=18,
        fontweight="bold",
        color="#2C3E50"
    )


    calc.text(
        0.05,
        0.75,
        f"p = {p:.2f}",
        fontsize=17,
        color="#2874A6"
    )


    calc.text(
        0.05,
        0.62,
        f"w = {w:.2f}",
        fontsize=17,
        color="#8E44AD"
    )


    calc.text(
        0.05,
        0.49,
        f"b = {b:.2f}",
        fontsize=17,
        color="#E67E22"
    )


    calc.text(
        0.05,
        0.34,
        f"n = wp + b = {n:.2f}",
        fontsize=17,
        color="#34495E"
    )


    calc.text(
        0.05,
        0.18,
        f"a = f(n) = f({n:.2f}) = {a:.2f}",
        fontsize=17,
        color="#27AE60",
        fontweight="bold"
    )


    plt.show()


# ============================================================
# LIVE OUTPUT
# ============================================================

output = widgets.interactive_output(
    draw_neuron,
    {
        'p': p_slider,
        'w': w_slider,
        'b': b_slider
    }
)


# ============================================================
# DISPLAY
# ============================================================

title = widgets.HTML(
    "<h2 style='color:#2C3E50;'>Neuron Controls</h2>"
)

controls = widgets.VBox([
    title,
    p_slider,
    w_slider,
    b_slider
])

display(
    widgets.HBox([
        controls,
        output
    ])
)
```

8. The actual output depends on the particular transfer function that is chosen. 
9. We will discuss transfer function in the next section
10. The bias is much like a weight, except that it has a constant input of 1.
11. However, if you do not want to have a bias in a particular neuron, it can be omitted. We will see examples of this in Chapter 3, 7 and 16
12. **Note :** *w* and *b* are both *adjustable* scalar parameters of the neurons.
13. Typically the *Transfer function* is chosen by the designer and then the parameters *w* and *b* will be adjusted by some learning rule
14. So that the neuron input/output relationship meets some specific goal (see Chapter 4 for an introduction to learning rules).
15. As described in the following section, we have different transfer functions for different purposes.
