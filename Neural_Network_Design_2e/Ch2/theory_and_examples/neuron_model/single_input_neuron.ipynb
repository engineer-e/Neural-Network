# Neuron Model

## Single-Input Neuron


> Weight, Bias, Net Input, Transfer Function

1. The scalar input _p_ is multiplied by the scalar weight _w_ to form _wp_, one of terms that is sent to the summar.
2. The other input, 1, is multiplied by a _bias b_ and then passed to the summer.
3. The Summer output _n_, often referred to as the _net input_, goes into a _transfer function f_, which produces the scalar neuron output _a_.
4. Some authors use the term
   - "activation function" rather than _transfer function_
   - "offset" rather than _bias_
5. This simple model back to the biological neuron that we discussed in _Chatper 1_
6. The weight _w_ corresponds to the strength of a synapse.
7. The Cell body is represented by the summation and the transfer function, and the neuron output _a_ represents the signal on the axon


$$\displaystyle a = f(wp+b)$$

```python
import ipywidgets as widgets
import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle, Circle, FancyArrowPatch
from IPython.display import display

# ============================================================
# SINGLE-INPUT NEURON
# n = wp + b
# a = f(n)
#
# Example transfer function:
# f(n) = n
# ============================================================

# ------------------------------------------------------------
# Transfer function
# ------------------------------------------------------------

def f(n):
    return n


# ------------------------------------------------------------
# Sliders
# ------------------------------------------------------------

p_slider = widgets.FloatSlider(
    value=1.0,
    min=-5.0,
    max=5.0,
    step=0.1,
    description='Input p:',
    continuous_update=True,
    style={'description_width': '80px'},
    layout=widgets.Layout(width='450px')
)

w_slider = widgets.FloatSlider(
    value=1.0,
    min=-5.0,
    max=5.0,
    step=0.1,
    description='Weight w:',
    continuous_update=True,
    style={'description_width': '80px'},
    layout=widgets.Layout(width='450px')
)

b_slider = widgets.FloatSlider(
    value=0.0,
    min=-5.0,
    max=5.0,
    step=0.1,
    description='Bias b:',
    continuous_update=True,
    style={'description_width': '80px'},
    layout=widgets.Layout(width='450px')
)


# ------------------------------------------------------------
# Text entry boxes
# ------------------------------------------------------------

p_text = widgets.FloatText(
    value=1.0,
    description='p =',
    layout=widgets.Layout(width='150px')
)

w_text = widgets.FloatText(
    value=1.0,
    description='w =',
    layout=widgets.Layout(width='150px')
)

b_text = widgets.FloatText(
    value=0.0,
    description='b =',
    layout=widgets.Layout(width='150px')
)


# Link slider and text box
widgets.link((p_slider, 'value'), (p_text, 'value'))
widgets.link((w_slider, 'value'), (w_text, 'value'))
widgets.link((b_slider, 'value'), (b_text, 'value'))


# ------------------------------------------------------------
# Draw the neuron
# ------------------------------------------------------------

def draw_neuron(p, w, b):

    # Calculate neuron
    n = w * p + b
    a = f(n)

    # --------------------------------------------------------
    # Figure
    # --------------------------------------------------------

    fig = plt.figure(figsize=(13, 7))
    fig.patch.set_facecolor("#F5F7FA")

    # ========================================================
    # NEURON DIAGRAM
    # ========================================================

    ax = fig.add_axes([0.03, 0.12, 0.65, 0.78])
    ax.set_xlim(0, 12)
    ax.set_ylim(0, 8)
    ax.axis("off")

    # --------------------------------------------------------
    # Title
    # --------------------------------------------------------

    ax.text(
        6, 7.5,
        "Single-Input Neuron",
        ha="center",
        fontsize=25,
        fontweight="bold",
        color="#17202A"
    )

    # --------------------------------------------------------
    # Input p
    # --------------------------------------------------------

    ax.text(
        1.0, 5.2,
        "Input",
        fontsize=17,
        fontweight="bold",
        color="#2874A6"
    )

    ax.add_patch(
        Circle(
            (1.5, 4.2),
            0.13,
            color="#2874A6"
        )
    )

    ax.text(
        1.15, 3.65,
        r"$p$",
        fontsize=25,
        color="#2874A6"
    )

    # Input arrow
    ax.add_patch(
        FancyArrowPatch(
            (1.65, 4.2),
            (3.2, 4.2),
            arrowstyle="-|>",
            mutation_scale=22,
            linewidth=2.5,
            color="#2874A6"
        )
    )

    # Weight
    ax.text(
        2.3, 4.65,
        r"$w$",
        fontsize=22,
        color="#8E44AD"
    )

    # --------------------------------------------------------
    # Summation block
    # --------------------------------------------------------

    sum_box = Rectangle(
        (3.2, 3.55),
        1.2,
        1.3,
        linewidth=3,
        edgecolor="#2C3E50",
        facecolor="white"
    )

    ax.add_patch(sum_box)

    ax.text(
        3.8, 4.2,
        r"$\Sigma$",
        ha="center",
        va="center",
        fontsize=36,
        color="#2C3E50"
    )

    # --------------------------------------------------------
    # Bias b
    # --------------------------------------------------------

    ax.add_patch(
        Circle(
            (3.8, 2.0),
            0.13,
            color="#E67E22"
        )
    )

    ax.add_patch(
        FancyArrowPatch(
            (3.8, 2.15),
            (3.8, 3.55),
            arrowstyle="-|>",
            mutation_scale=20,
            linewidth=2.5,
            color="#E67E22"
        )
    )

    ax.text(
        4.15, 2.55,
        r"$b$",
        fontsize=22,
        color="#E67E22"
    )

    # --------------------------------------------------------
    # n = wp+b
    # --------------------------------------------------------

    ax.add_patch(
        FancyArrowPatch(
            (4.4, 4.2),
            (5.7, 4.2),
            arrowstyle="-|>",
            mutation_scale=22,
            linewidth=2.5,
            color="#34495E"
        )
    )

    ax.text(
        5.05, 4.65,
        r"$n$",
        fontsize=22,
        color="#34495E"
    )

    # --------------------------------------------------------
    # Transfer function f
    # --------------------------------------------------------

    f_box = Rectangle(
        (5.7, 3.55),
        1.3,
        1.3,
        linewidth=3,
        edgecolor="#27AE60",
        facecolor="#EAF8F0"
    )

    ax.add_patch(f_box)

    ax.text(
        6.35, 4.2,
        r"$f$",
        ha="center",
        va="center",
        fontsize=34,
        color="#27AE60"
    )

    # --------------------------------------------------------
    # Output a
    # --------------------------------------------------------

    ax.add_patch(
        FancyArrowPatch(
            (7.0, 4.2),
            (8.6, 4.2),
            arrowstyle="-|>",
            mutation_scale=22,
            linewidth=2.5,
            color="#C0392B"
        )
    )

    ax.text(
        9.1, 4.2,
        r"$a$",
        fontsize=26,
        fontweight="bold",
        color="#C0392B"
    )

    # --------------------------------------------------------
    # Main equation
    # --------------------------------------------------------

    ax.text(
        6.0, 1.25,
        r"$a=f(wp+b)=f(n)$",
        ha="center",
        fontsize=23,
        color="#17202A"
    )

    # ========================================================
    # LIVE CALCULATION PANEL
    # ========================================================

    calc = fig.add_axes([0.71, 0.45, 0.25, 0.35])
    calc.axis("off")

    calc.text(
        0.5, 0.95,
        "Live Calculation",
        ha="center",
        fontsize=18,
        fontweight="bold",
        color="#2C3E50"
    )

    calc.text(
        0.05, 0.75,
        rf"$p={p:.2f}$",
        fontsize=17,
        color="#2874A6"
    )

    calc.text(
        0.05, 0.62,
        rf"$w={w:.2f}$",
        fontsize=17,
        color="#8E44AD"
    )

    calc.text(
        0.05, 0.49,
        rf"$b={b:.2f}$",
        fontsize=17,
        color="#E67E22"
    )

    calc.text(
        0.05, 0.32,
        rf"$n=wp+b={n:.2f}$",
        fontsize=17,
        color="#34495E"
    )

    calc.text(
        0.05, 0.16,
        rf"$a=f(n)=f({n:.2f})={a:.2f}$",
        fontsize=17,
        color="#27AE60",
        fontweight="bold"
    )

    # ========================================================
    # FUNCTION EXAMPLE
    # ========================================================

    fig.text(
        0.72, 0.25,
        r"Example: $f(n)=n$",
        fontsize=17,
        color="#2C3E50"
    )

    fig.text(
        0.72, 0.20,
        r"$f(4.5)=4.5$",
        fontsize=17,
        color="#27AE60"
    )

    fig.text(
        0.72, 0.14,
        "Change p, w or b\n"
        "using the sliders.",
        fontsize=12,
        color="#7F8C8D"
    )

    plt.show()


# ------------------------------------------------------------
# Live output
# ------------------------------------------------------------

output = widgets.interactive_output(
    draw_neuron,
    {
        "p": p_slider,
        "w": w_slider,
        "b": b_slider
    }
)


# ------------------------------------------------------------
# Control panel
# ------------------------------------------------------------

title = widgets.HTML(
    "<h2 style='color:#2C3E50;'>Neuron Inputs</h2>"
)

controls = widgets.VBox([
    title,

    p_slider,
    p_text,

    w_slider,
    w_text,

    b_slider,
    b_text
])

display(controls)
display(output)
```

8. The actual output depends on the particular transfer function that is chosen.

10. The bias is much like a weight, except that it has a constant input of 1.
11. However, if you do not want to have a bias in a particular neuron, it can be omitted. We will see examples of this in Chapter 3, 7 and 16
12. **Note :** *w* and *b* are both *adjustable* scalar parameters of the neurons.
13. Typically the *Transfer function* is chosen by the designer and then the parameters *w* and *b* will be adjusted by some learning rule
14. So that the neuron input/output relationship meets some specific goal (see Chapter 4 for an introduction to learning rules).
