### Layer Superscript


1. Now consider a network with several layers.
2. Each layer has its own weight matrix **W**, its own bias vector **b**, a net input vector **n** and an output vector **a**. 
3. We need to introduce some additional notation to distinguish between these layers.
4. We will use superscripts to identify the layers. 
5. Specifically, we append the number of the layer as a *superscript* to the names for each of these variables.
6. Thus, the weight matrix for the first layer is written as $W^1$, and the weight matrix for the second layer is written as $W^2$.
7. This notation is used in the three-layer network shown in Figure 2.9.
8. As shown, there are **R** inputs, $S^1$ neurons in the first layer, $S^2$ neurons in the second layer, etc. As noted, different layers can have different numbers of neurons.
9. The outputs of layers one and two are the inputs for layers two and three.
10. Thus layer 2 can be viewed as a one-layer network with $R = S^1$ inputs, $S=S^2$ neurons, and an $S^2 × S^1$ weight matrix $W^2$.
11. The input to layer 2 is $a^1$, and the output is $a^2$.

### Output Layer Hidden Layers

1. A layer whose output is the network output is called an *output layer*.
2. The other layers are called *hidden layers*.
3. The network shown above has an output layer (layer 3) and two hidden layers (layers 1 and 2).
4. The same three-layer network discussed previously also can be drawn using our abbreviated notation, as shown in Figure 2.10.
5. Multilayer networks are more powerful than single-layer networks. 
6. For instance, a two-layer network having a sigmoid first layer and a linear second layer can be trained to approximate most functions arbitrarily well.
7. Single layer networks cannnot do this.
8. At this point the number of choices to be made in specifying a network may look overwhelming, so let us consider this topic.
9. The problem is not as bad as it looks. 
10. First, recall that the number of inputs to the network and the number of outputs from the network are defined by external problem specifications.
11. So if there are four external variables to be used as inputs, there are four inputs to the network.
12. Similarly, if there are to be seven outputs from the network, there must be seven neurons in the output layer.
13. Finally, the desired characteristics of the output signal also help to select the transfer function for the output layer.
14. If an output is to be either -1 or 1, then a symmetrical hard limit transfer function should be used. 
15. Thus, the architecture of a single-layer network is almost completely determined by problem specifications, including the specific number of inputs and outputs and the particular output signal characteristic.
16. Now, what if we have more than two layers?
17. Here the external problem does not tell you directly the number of neurons required in the hidden layer.
18. In fact, there are few problems for which one can predict the optimal number of neurons needed in a hidden layer.
19. This problem is an active area of research. 
20. We will develop some feeling on this matter as we proceed to Chapter 11, Backpropagation.
21. As for the number of layers, most practical neural networks have just two or three layers.
22. Four or more layers are used rarely.
23. We should say something about the use of biases. 
24. One can choose neurons with or without biases.
25. The bias gives the network an extra variable, and so you might expect that networks with biases would be more powerful than those without, and that is true.
26. Note, for instance that a neuron without a bias will always have a net input *n* of zero when the network inputs **p** are zero. 
27. This may not be desirable and can be avoided by the use of a bias.
28. The effect of the bias is discussed more fully is Chapters 3,4 and 5. 
29. In later chapters we will omit a bias is some examples or demonstrations.
30. In some cases this is done simply to reduce the number of network parameters. 
31. With just two variables, we can plot system convergence in a two-dimensional plane.
32. Three or more variables are difficult to display.

### Recurrent Networks

#### Delay

1. Before we discuss recurrent networks, we need to introduce some simple building blocks.
2. The first is the *delay* block, which is illustrated in Figure 2.11.
3. The delay output **a(t)** is computed from its input **u**(t) according to 
$$a(t) = u(t-1)$$
4. Thus the output is the input delayed by one time step. (This assumes that time is updated in discreate steps and takes on only integer values.) Eq. (2.7) requires that the output be initialized at time *t* = 0. 
5. This initial condition is indicated in Figure 2.11  by the arrow coming into the bottom of the delay block. 


#### Integrator

1. Another related building block, which we will use for the continuous-time recurrent networks in Chapters 18-21, is the *integrator*, which is shown in Figure 2.12.
2. The integrator output a(t) is computed from its input u(t) according to 
$$\displaystyle a(t) = \int_0^t u(τ) dτ + a(0) $$
3. The initial condition a(0) is indicated by the arrow coming into the bottom of the integrator block.


#### Recurrent Network

1. We are now ready to introduce recurrent networks.
2. A *recurrent network* is a network with feedbadk; some of its outputs are connected to its inputs. 
3. This is quite different from the networks that we have studied thus far, which were strictly feedforward with no backward connections.
4. One type of discrete-time recurrent network is shown in Figure 2.13. 
5. In this particular network the vector **p** supplies the initial conditions (i.e., a(0) = p).
6. Then future outputs of the network are computed from previous outputs 
$$\displaystyle a(1) = satlins(Wa(0)+b), a(2) = satlins(Wa(1)+b), \cdots$$
7. Recurrent networks are potentially more powerful than feedforward networks and can exhibit temporal behavior. 
8. These types of networks are discussed in Chapter 3, 14 and 18-21
