# Network Architecture

## [A Layer of Neurons](https://engineer-e.github.io/Neural-Network/Neural_Network_Design_2e/Ch2/theory_and_examples/network_architectures/a_layer_of_neurons.html)

1. A single-*layer* network of *S* neurons is shown in Figure 2.7. 
2. Note that each of the *R* inputs is connected to each of the neurons and that the weight matrix now has *S* rows



3. The layer includes the weight matrix, the summers, the bias vector **b**, the transfer function boxes adn the output vector **a**.
4. Some authors refer to the inputs as another layer, but we will not do that here
5. Each elements of the input vector **p** is connected to each neuron through the weight matrix **W**.
6. Each neuron has a bias $b_i$, a summer, a transfer function *f* and an output $a_i$.
7. Taken together, the outputs form the output vector **a**
8. It is common for the number of inputs to a layer to be different from the number of neurons (i.e., *R* ≠ *S*).
9. You might ask if all the neurons in a layer must have the same transfer function.
10. The answer is no; you can define a single (composite) layer of neurons having different transfer functions by combining two of the networks shown above in parallel. 
11. Both networks would have the same inputs, and each network would create some of the outputs.
12. The input vector elements enter the network through the weight matrix **W**:
$$W = \begin{bmatrix*}[r]
   W_{1,1} & W_{1,2} & \cdots &W_{1,R} \\
   W_{2,1} & W_{1,2} & \cdots &W_{1,R} \\
   \vdots &  \vdots &  & \vdots \\
   W_{S,1} & W_{S,2} & \cdots &W_{S,R} \\
\end{bmatrix*} $$
13. As noted previously, the row indices of the elements of matrix **W** indicate the destination neuron associated with that weight, while the column indices indicate the source of the input for that weight. 
14. Thus, the indices in $w_{3,1}$ say that this weight represents the connection *to* the third neuron *from* the second source. 
15. Fortunately, the *S*-neuron, *R*-input, one-layer network also can be drawn in abbreviated notation, as shown in **Figure 2.8.**



16. Here again, the symbols below the variables tell you that for this layer, **p** is a vector of length *R*, **W** in an $S × R$ matrix, and **a** & **b** are vectors of length *S*. 
17. As defined previously, the layer includes the weight matrix, the summation and multiplication operations, the bias vector **b**, the transfer function boxes and the output vector.
